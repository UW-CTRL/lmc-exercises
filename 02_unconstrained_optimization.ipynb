## Rosenbrock function


The Rosenbrock function is a classic test problem for optimization algorithms. It is defined as:

$$
f(x, y) = (a - x)^2 + b(y - x^2)^2
$$

where typically \( a = 1 \) and \( b = 100 \).

```python
def rosenbrock_func(x, a=1, b=100):
    return (a - x[0])**2 + b * (x[1] - x[0]**2)**2
```

- **Step size (learning rate)**:
  - Too large: iterates can overshoot the narrow valley, oscillate, or diverge.
  - Too small: convergence is very slow and may stall near flat/saddle regions; more steps and numerical noise dominate.
  - Practical fix: use an adaptive/decayed rate or a line search.

- **Dependence on initial conditions**:
  - The Rosenbrock function has a single global minimum at (1, 1), but it is non-convex and has saddles.
  - Many initial points will converge to (1, 1) if the step size is stable; however, some starts combined with a poor step size can get trapped near saddles, zig-zag for a long time, or diverge.
  - So, convergence speed and even success depend on both the start and the step size schedule.

- **Effect of the Rosenbrock geometry**:
  - The banana-shaped, curved, and very narrow valley makes the problem ill-conditioned.
  - Gradients point steeply across the valley but weakly along it, causing zig-zagging and slow progress with plain gradient descent.
  - Methods with momentum, preconditioning, or second-order information typically handle this curvature much better.

## Himmelblau’s function

Himmelblau’s function is a well-known test function for optimization algorithms, particularly in two dimensions. It is defined as:

$$
f(x, y) = (x^2 + y - 11)^2 + (x + y^2 - 7)^2
$$

This function is notable for having multiple local minima and saddle points, making it a challenging landscape for optimization methods. The global minima are located at four points in the $(x, y)$ plane:

- $(3.0, 2.0)$
- $(-2.805, 3.131)$
- $(-3.779, -3.283)$
- $(3.584, -1.848)$

At each of these points, the function value is zero: $f(x, y) = 0$. The complex topology of Himmelblau’s function makes it useful for evaluating the performance and robustness of optimization algorithms.

```python
def himmelblau_func(x):
    return (x[0]**2 + x[1] - 11)**2 + (x[0] + x[1]**2 - 7)**2
```

$$
f(x, y) = a(y - bx^2 + cx - r)^2 + s(1 - t)\cos(x) + s
$$

where the typical parameter values are:
- $ a = 1 $
- $ b = \frac{5.1}{4\pi^2} $
- $ c = \frac{5}{\pi} $
- $ r = 6 $
- $ s = 10 $
- $ t = \frac{1}{8\pi} $

The function is usually evaluated on the domain $ x \in [-5, 10] $ and $ y \in [0, 15] $.

The Branin function has three global minima, making it useful for testing the performance of global optimization algorithms. Its landscape is characterized by multiple valleys and ridges, which can challenge optimization methods.
