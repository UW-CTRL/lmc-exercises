# Infinite horizon LQR

## 1. Discrete-time infinite horizon LQR

We consider the same discrete-time double integrator model you saw before. The state is
$$
x_k = \begin{bmatrix} p_k \\ v_k \end{bmatrix},
$$
where $p_k$ is position and $v_k$ is velocity.

The input $u_k$ is an acceleration command. The time-discretized dynamics are
$$
x_{k+1} = A x_k + B u_k.
$$

We'll use a quadratic cost of the form
$$
J = \sum_{k=0}^{N-1} (x_k^\top Q x_k + u_k^\top R u_k) + x_N^\top Q_f x_N.
$$

In finite-horizon LQR, the optimal controller is time-varying: $u_k = -K_k x_k$.
In infinite-horizon LQR, the optimal controller becomes time-invariant: $u = -K_\infty x$.

We'll see both.
